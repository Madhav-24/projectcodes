## Data Exploration

In this section, we will explore the dataset, visualize some samples, and understand the distribution of PPE compliance among workers.

```python
# Function to visualize a batch of images
def visualize_images(images, titles=None):
    n = len(images)
    plt.figure(figsize=(15, 5))
    for i in range(n):
        plt.subplot(1, n, i + 1)
        plt.imshow(images[i])
        plt.axis('off')
        if titles:
            plt.title(titles[i])
    plt.show()

# Example usage of visualize_images function
# Assuming we have a list of images loaded in a variable called `image_list`
# visualize_images(image_list)
```
